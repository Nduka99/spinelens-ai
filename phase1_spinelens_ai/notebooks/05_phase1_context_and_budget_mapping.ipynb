# Phase 1 - Context, Scope and Reading Guide

SpineLens AI is the Phase 1 evidence layer for the Innovation Spine "Make It Visible" package
(GBP 1,000,000). It answers one sharp question with open data and transparent modelling:

> Where does the city-to-B-KQ journey become illegible, and what low-cost interventions create
> the highest route clarity per pound?

Everything is built notebooks-first, with a forensic evidence protocol: no output is treated as
funding-grade until it passes evidence gates. Evidence levels: 0 not tested, 1 reachable,
2 raw, 3 audited, 4 cross-checked, 5 field-validated. **Costs are Level 1 indicative pending a
quantity-surveyor costing; nothing here is a funding-facing claim.**

## Evidence-gate sequence

- **G0** study boundary + acquisition ledger
- **G0A** route intent + gateway definition
- **G0B** raw acquisition + quality audit (OSM walk network, OS Greenspace)
- **G0C** crossing evidence + options appraisal (DfT AADF + STATS19)
- **G0D** pavilion suitability + site risk
- **G0E** authoritative cross-check (DfT road class) + HM Land Registry parcel screen

## The model in one line

City-core approaches (Colmore, New Street, Moor Street, Snow Hill) follow a tactical amber
corridor + wayfinders into the **Ryder Street gateway pavilion** (a wooden, reversible, public
information hub), then onward via the Aston green park. The **Nechells / north** side crosses the
Dartmouth barrier and is served by the crossing + wayfinders (no second pavilion).
